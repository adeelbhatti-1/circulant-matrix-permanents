```python
import itertools
from sympy import symbols, Matrix, zeros, simplify
import numpy as np

def permanent_ryser(matrix):
    """
    Calculate the permanent using Ryser's formula (SymPy compatible version).
    matrix: SymPy Matrix
    """
    n = matrix.rows
    total = 0

    # Iterate over all subsets of columns (except empty set)
    for k in range(1, n+1):
        for cols in itertools.combinations(range(n), k):
            product = 1
            for row in range(n):
                row_sum = sum(matrix[row, col] for col in cols)
                product *= row_sum
            total += (-1)**(n - k) * product

    return total

def create_circulant_C(n, c1, c2, c3):
    """
    Create circulant matrix C with diagonal c1 and off-diagonals c2, c3.
    Returns a SymPy Matrix.
    """
    C = zeros(n, n)
    for i in range(n):
        C[i, i] = c1  # Main diagonal
        C[i, (i+1)%n] = c2  # Next diagonal
        C[i, (i+2)%n] = c3  # Next next diagonal
    return C

def permanent_circulant_3(n, c1, c2, c3):
    """
    Calculate permanent of circulant matrix C with 3 non-zero entries using the given formulas.
    Returns a symbolic expression.
    """
    if n == 0:
        return 1

    # Base terms
    base_terms = c1**n + c2**n + c3**n

    # Initialize sum for the remaining terms
    sum_terms = 0

    # Determine the range for r based on whether n is even or odd
    max_r = n//2 if n % 2 == 0 else (n-1)//2

    # Iterate over all r and alpha
    for r in range(1, max_r + 1):
        # All combinations of alpha with size r
        for alpha in itertools.combinations(range(n), r):
            # Create submatrix B(alpha|alpha)
            keep = [i for i in range(n) if i not in alpha]
            if not keep:
                per_sub = 1  # Permanent of empty matrix is 1
            else:
                # Create the submatrix as a SymPy matrix
                submatrix = zeros(len(keep), len(keep))
                for i, row in enumerate(keep):
                    for j, col in enumerate(keep):
                        offset = (col - row) % n
                        if offset == 1:
                            submatrix[i, j] = c2
                        elif offset == 2:
                            submatrix[i, j] = c3
                        # else remains 0

                # Calculate permanent of submatrix using Ryser's formula
                per_sub = permanent_ryser(submatrix)

            # Add term to sum (c1^r * per(B(α|α)))
            sum_terms += (c1 ** r) * per_sub

    return simplify(base_terms + sum_terms)

# Example usage with symbolic variables
if __name__ == "__main__":
  n=8
  c1, c2, c3 = symbols('c1 c2 c3')

  # Create the circulant matrix
  C = create_circulant_C(n, c1, c2, c3)
  print("\nCirculant matrix C:")
  print(C)

  # Calculate permanent using our formula
  per = permanent_circulant_3(n, c1, c2, c3)
  print(f"\nPermanent of C (using formula):")
  print(per)

  # Verification using Ryser's formula directly on C
  per_ryser = permanent_ryser(C)
  print(f"\nPermanent of C (using Ryser directly):")
  print(simplify(per_ryser))
```
